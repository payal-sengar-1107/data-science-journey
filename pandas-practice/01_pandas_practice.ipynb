# importing pandas

```python
import numpy as np
import pandas as pd
```

# series from lists

```python
country = ['India','pakistan','USA','Nepal','Shrilanka']

pd.Series(country)
```

```python
# integers
runs = [13,24,56,78,100]

pd.Series(runs)
```

```python
# custom index

marks = [67,57,89,100]
subjects = ['Math','English','Science','Hindi']

pd.Series(marks)
pd.Series(subjects)
```

```python
# custom index

marks = [67,57,89,100]
subjects = ['Math','English','Science','Hindi']

pd.Series(marks,index=subjects)
```

```python
# setting a name
pd.Series(marks,index=subjects,name='Pyson ke marks')
```

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='radha ke marks')
```

# Series from dict

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}
marks_series = pd.Series(marks)
marks_series
```

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}
marks_series = pd.Series(marks,name='radha ke makrs')
marks_series
```
